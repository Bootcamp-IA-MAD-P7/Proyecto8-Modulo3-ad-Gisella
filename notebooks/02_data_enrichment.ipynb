```python
# Cargar el dataset limpio de Airbnb
df = pd.read_csv('../data/processed/madrid_airbnb_clean.csv')

# Cargar las 5 fuentes de enriquecimiento
poblacion = pd.read_csv('../data/raw/poblacion_distritos.csv')
renta = pd.read_csv('../data/raw/renta_distritos.csv')
alquiler = pd.read_csv('../data/raw/alquiler_referencia_distritos.csv')
eventos = pd.read_csv('../data/raw/eventos_distritos.csv')
vut = pd.read_excel('../data/raw/VIVIENDAS_USO_TURISTICO.xlsx')

print(df.shape)
print(poblacion.shape, renta.shape, alquiler.shape, eventos.shape, vut.shape)
```

```python
import re

# Lista de referencia: los 21 distritos tal y como están en el dataset de Airbnb (no se toca)
distritos_referencia = sorted(df['neighbourhood_group'].unique())

def normalizar_distrito(nombre):
    nombre = str(nombre).strip()
    nombre = nombre.replace('Distrito de ', '')
    nombre = re.sub(r'\s+', ' ', nombre)          # colapsa espacios múltiples
    nombre = re.sub(r'\s*-\s*', ' - ', nombre)     # guion siempre con espacios: " - "
    nombre = nombre.replace('Tetuan', 'Tetuán')    # corrige tilde que falta en el VUT
    return nombre.strip()

# Aplicar a las 4 fuentes propias
poblacion['distrito'] = poblacion['distrito'].apply(normalizar_distrito)
renta['distrito'] = renta['distrito'].apply(normalizar_distrito)
alquiler['distrito'] = alquiler['distrito'].apply(normalizar_distrito)
eventos['distrito'] = eventos['distrito'].apply(normalizar_distrito)

# Aplicar al VUT
vut['distrito'] = vut['DISTRITO'].apply(normalizar_distrito)

# Comprobar que todo cuadra con los 21 distritos de referencia
print("Referencia (21):", len(distritos_referencia))
print("Población OK:", set(poblacion['distrito']) == set(distritos_referencia))
print("Renta OK:", set(renta['distrito']) == set(distritos_referencia))
print("Alquiler OK:", set(alquiler['distrito']) == set(distritos_referencia))
print("Eventos OK:", set(eventos['distrito']) == set(distritos_referencia))
print("VUT distritos únicos tras normalizar:", sorted(vut['distrito'].unique()))
print("VUT OK:", set(vut['distrito']) == set(distritos_referencia))
```

```python
def normalizar_distrito(nombre):
    nombre = str(nombre).strip()
    nombre = nombre.replace('Distrito de ', '')
    nombre = re.sub(r'\s+', ' ', nombre)
    nombre = re.sub(r'\s*-\s*', ' - ', nombre)
    # Correcciones de tildes que faltan en el VUT
    nombre = nombre.replace('Tetuan', 'Tetuán')
    nombre = nombre.replace('Chamartin', 'Chamartín')
    nombre = nombre.replace('Chamberi', 'Chamberí')
    return nombre.strip()

# Volver a aplicar solo al VUT (las otras 4 ya salieron OK)
vut['distrito'] = vut['DISTRITO'].apply(normalizar_distrito)

print("VUT distritos únicos tras normalizar:", sorted(vut['distrito'].unique()))
print("VUT OK:", set(vut['distrito']) == set(distritos_referencia))
```

```python
vut_por_distrito = vut.groupby('distrito')['UNIDADES_VUT'].sum().reset_index()
vut_por_distrito.columns = ['distrito', 'total_vut']
print(vut_por_distrito.shape)
vut_por_distrito
```

```python
df_enriched = df.merge(poblacion, left_on='neighbourhood_group', right_on='distrito', how='left') \
                .merge(renta, on='distrito', how='left') \
                .merge(alquiler, on='distrito', how='left') \
                .merge(eventos, on='distrito', how='left') \
                .merge(vut_por_distrito, on='distrito', how='left')

# Eliminar columnas 'distrito' duplicadas si sobran (nos quedamos con neighbourhood_group como clave original)
df_enriched = df_enriched.drop(columns=['distrito'])

print(df_enriched.shape)
df_enriched.head()
```

```python
# Comprobar que no falte ningún dato tras el merge
columnas_nuevas = ['poblacion_total', 'renta_media', 'precio_alquiler', 'evento', 'total_vut']
print(df_enriched[columnas_nuevas].isnull().sum())
```

## Resumen del enriquecimiento de datos

Este notebook enriquece el dataset limpio de Airbnb Madrid con cinco variables a nivel de distrito, obtenidas del Ayuntamiento de Madrid / INE, usando `neighbourhood_group` como clave de unión.

**Fuentes de enriquecimiento:**

| Variable | Descripción | Fuente |
|---|---|---|
| `poblacion_total` | Población total por distrito (Padrón Municipal) | Ayuntamiento de Madrid |
| `renta_media_hogar` | Renta media por hogar y distrito | Ayuntamiento de Madrid / INE Atlas de renta |
| `alquiler_medio_mes` | Precio de alquiler de referencia por distrito | Ayuntamiento de Madrid |
| `tiene_gran_recinto` | Indicador binario: si el distrito tiene un gran recinto de eventos | Elaboración propia (pandas) |
| `total_vut` | Total de viviendas de uso turístico (VUT) con licencia por distrito | Geoportal Ayuntamiento de Madrid |

**Pasos principales:**
1. Carga del dataset limpio de Airbnb (`madrid_airbnb_clean.csv`) y de las cinco fuentes de enriquecimiento.
2. Normalización de los nombres de distrito entre fuentes — el dataset VUT requería eliminar el prefijo `"Distrito de "`, quitar espacios sobrantes, unificar el formato del guion y corregir tildes faltantes (Tetuán, Chamartín, Chamberí).
3. Agregación del VUT por distrito, sumando `UNIDADES_VUT` en `total_vut`.
4. Merge tipo `left` de las cinco fuentes sobre el dataset original usando `neighbourhood_group` como clave, conservando las 19.618 filas originales.
5. Verificación de que no se introdujeron valores nulos en ninguna de las cinco columnas nuevas.

**Resultado:** `df_enriched` — 19.618 filas × 21 columnas, guardado en `../data/processed/madrid_airbnb_enriched.csv`. Las columnas originales de Airbnb no fueron modificadas, conforme a los requisitos del proyecto.
